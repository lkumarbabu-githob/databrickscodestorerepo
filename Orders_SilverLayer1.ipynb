```python
%sql
CREATE OR REPLACE TEMP VIEW vwPrjOrdertrimmed as
select trim(order_id) as order_id,
trim(customer_id) as customer_id,
trim(product_id) as product_id,
trim(order_date) as order_date,
trim(quantity) as quantity,
trim(total_amount) as total_amount
from sparkschemaprojectsales.tblprjorderssource
```

```python
%sql
create or replace temp view vwprjorderstandardized as
select 
upper(order_id) as order_id,
upper(customer_id) as customer_id,
upper(product_id) as product_id,
order_date,
cast(quantity as int) as quantity,
cast(total_amount as double) as total_amount
from vwprojordersdistinct
```

```python
%sql
create or replace temp view vwprojordersdist as
select distinct order_id,customer_id,product_id,order_date,quantity,total_amount
from vwprjorderstandardized
    
```

```python
%sql
select order_id,customer_id,product_id,order_date,quantity,total_amount
from vwprojordersdist
where order_id is null or customer_id is null or product_id is null or order_date is null or quantity is null or total_amount is null
```

```python
%sql
select * from vwprojordersdist
```

```python
%sql
create or replace temp view vwprjordervalidorderid as
select * from vwprojordersdist where order_id rlike '^O[0-9]{5}$'
```

```python
%sql
create or replace temp view vwprjordervalidcustomerid as
select * from vwprjordervalidorderid where customer_id rlike '^C[0-9]{5}$'
```

```python
%sql
create or replace temp view vwprjordervalidproductid as
select * from vwprjordervalidcustomerid where product_id rlike '^P[0-9]{4}$'
```

```python
%sql
select order_id,count(*) as cnt from vwprjordervalidproductid
group by order_id
having count(*)  > 1
order by cnt desc
```

```python
%sql
create or replace temp view vwprjordersorderidunique as
select * from 
(select *, row_number() over(partition by order_id order by order_date ) as rn from vwprjordervalidproductid)
duplicatecheck 
where rn = 1
```

```python
%sql
create or replace temp view vwprjorderqtyvalidate as
select * from vwprjordersorderidunique where quantity is null or quantity > 0
    
```

```python
%sql
create or replace temp view vwprjordertotalamtvalidate as
select order_id,customer_id,product_id,order_date,quantity,
round(total_amount,2) as total_amount 
from vwprjorderqtyvalidate where total_amount is null or total_amount > 0
    
```

```python
%sql
select * from vwprjordertotalamtvalidate where order_date is null or order_date not rlike '^[0-9]{4}-[0-9]{2}-[0-9]{2}$' or order_date > current_date()
    
```

```python
%sql
create or replace temp view vwprjorderdatevalidate as
select * from vwprjordertotalamtvalidate where order_date <= current_date()
```

```python
%sql
create or replace temp view vwprjorderdatederivedcoloumns as
select order_id,customer_id,product_id,order_date,quantity,
total_amount,
date_format(order_date,'yyyy') as order_year,
date_format(order_date,'MM') as order_month,
quarter(order_date) as order_quarter,
date_format(order_date,'d') as order_day
from vwprjorderdatevalidate 
```

```python
%sql
create or replace temp view vwprjordersavgvalue as
select *, round(total_amount/quantity,2) as avg_value
from vwprjorderdatederivedcoloumns
    
```

```python
%sql
create or replace temp view vwprjordervalueclassified as
select *,case when total_amount < 1000 then 'low'
when total_amount >= 1000 and total_amount < 5000 then 'medium'
else 'high'
end as value_class
from vwprjordersavgvalue
    
```

```python
%sql
create or replace temp view vwprjordersfinal as
select *,
current_timestamp() as created_timesatmp,
'orderssfile.parquet' as sourcefile,
'SILVER' AS datalayer
from vwprjordervalueclassified
```

```python
%sql
create or replace table tblPrjOrderSilver 
using delta
as
select * from vwprjordersfinal
```
