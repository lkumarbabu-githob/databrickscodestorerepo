```sql
CREATE TABLE IF NOT EXISTS dim_orders
(
    order_key   BIGINT NOT NULL, --SURROGATE KEY
    order_id    STRING NOT NULL,
    customer_id STRING,
    product_id	string,
    order_date	string,
    quantity	int,
    total_amount decimal(18,2),
    order_year	int	,
    order_month	int	,
    order_quarter	int	,
    order_day	int	,
    avg_item_value	decimal(18,2),
    order_value_category string,
    source_created_datetime TIMESTAMP,
    source_file      STRING,
    gold_created_datetime TIMESTAMP NOT NULL,
    gold_updated_datetime TIMESTAMP NOT NULL
) USING DELTA
```

```sql
CREATE OR REPLACE TEMP VIEW orders_gold_source AS
SELECT
    XXHASH64(order_id) AS order_key,
    order_id,
    customer_id,
    product_id,
    order_date,
    quantity,
    CAST(total_amount as decimal(18,2)) as total_amount,
    order_year,
    order_month	,
    order_quarter,
    order_dayofmonth as order_day,
    CAST(avg_value AS decimal(18,2)) AS avg_item_value,
    value_class  as order_value_category,
    sourcefile as source_file,
    current_timestamp()  AS load_datetime
FROM
(
    SELECT * , ROW_NUMBER() OVER(PARTITION BY order_id ORDER BY created_timesatmp DESC) 
    AS
    row_num FROM sliverschemasales.tblprjordersilver
) src WHERE row_num = 1
```

```sql
select * from orders_gold_source
```

```sql
MERGE INTO dim_orders AS target
USING orders_gold_source AS source
ON target.order_id = source.order_id
WHEN MATCHED AND
(
        NOT (target.customer_id          <=> source.customer_id)
    OR NOT (target.product_id           <=> source.product_id)
    OR NOT (target.order_date           <=> source.order_date)
    OR NOT (target.quantity             <=> source.quantity)
    OR NOT (target.total_amount         <=> source.total_amount)
    OR NOT (target.order_year           <=> source.order_year)
    OR NOT (target.order_month          <=> source.order_month)
    OR NOT (target.order_quarter        <=> source.order_quarter)
    OR NOT (target.order_day            <=> source.order_day)
    OR NOT (target.avg_item_value       <=> source.avg_item_value)
    OR NOT (target.order_value_category <=> source.order_value_category)
    OR NOT (target.source_file          <=> source.source_file)
)
THEN UPDATE SET
    TARGET.customer_id          = source.customer_id,
    TARGET.product_id           = source.product_id,
    TARGET.order_date           = source.order_date,
    TARGET.quantity             = source.quantity,
    TARGET.total_amount         = source.total_amount,
    TARGET.order_year           = source.order_year,
    TARGET.order_month          = source.order_month,
    TARGET.order_quarter        = source.order_quarter,
    TARGET.order_day            = source.order_day,
    TARGET.avg_item_value       = source.avg_item_value,
    TARGET.order_value_category = source.order_value_category,
    TARGET.source_file          = source.source_file,
    TARGET.gold_updated_datetime = source.load_datetime
WHEN NOT MATCHED
THEN INSERT
(
    order_key,
    order_id,
    customer_id,
    product_id,
    order_date,
    quantity,
    total_amount,
    order_year,
    order_month,
    order_quarter,
    order_day,
    avg_item_value,
    order_value_category,
    source_file,
    gold_created_datetime,
    gold_updated_datetime
)
VALUES
(
    source.order_key,
    source.order_id,
    source.customer_id,
    source.product_id,
    source.order_date,
    source.quantity,
    source.total_amount,
    source.order_year,
    source.order_month,
    source.order_quarter,
    source.order_day,
    source.avg_item_value,
    source.order_value_category,
    source.source_file,
    source.load_datetime,
    source.load_datetime
)
```

```sql
CREATE OR REPLACE VIEW vw_monthly_sales AS
SELECT
    order_year,
    order_month,
    COUNT(DISTINCT order_id) as total_orders,
    SUM(quantity) as total_quantity,
    ROUND(SUM(total_amount), 2) as total_sales,
    ROUND(AVG(total_amount), 2) as avg_order_value
FROM dim_orders
GROUP BY order_year,order_month
```

```sql
select * from vw_monthly_sales order by order_year,order_month
```

```sql
CREATE OR REPLACE VIEW vw_sales_by_category AS
SELECT
    order_value_category,
    COUNT(DISTINCT order_id) as total_orders,
    SUM(quantity) as total_quantity,
    ROUND(SUM(total_amount), 2) as total_sales
FROM dim_orders
GROUP BY order_value_category
```
